### Buckley--Leverett PINN with Entropy Condition

The Buckley--Leverett equation constrained by the Oleinik entropy condition, for immiscible, isothermal, horizontal two-phase flow, with the conservation equation for the wetting-phase saturation $S_{w}$, is defined as

$$
\phi \frac{\partial S_{w}}{\partial t} + v \frac{\partial f(S_{w})}{\partial x} = 0, \qquad (x,t)\in[0,1],
$$

subject to the conditions

$$ S(0,t) = 1 - S_{nr}, $$

as the boundary condition, and

$$ S(x,0) = S_{wr}, $$

as the initial condition.

The entropy constraint, specifically the Oleinik condition, retains $f(\hat{S}_{w})$ for the simulation of the rarefaction part of the solution and is defined as

$$
f_{w}(\hat{S}_{w}) =
\begin{cases}
f(S_{wr}) + \sigma(\hat{S}_{w} - S_{wr}), & \hat{S}_{w} < S_{wf}, \\[4pt]
f(\hat{S}_{w}), & \hat{S}_{w} \geq S_{wf}.
\end{cases}
$$

where $\sigma = \frac{dx}{dt}$ is the characteristic velocity of $S_{wf}$ (the wetting-saturation front) throughout the domain $(x,t)$.

For this approximation, the loss-function model is constrained to the physics using the residual form of the governing equation together with the boundary- and initial-condition terms:

$$\mathcal{L} = \mathcal{L}_{physics}+\mathcal{L}_{\mathcal{I}}+\mathcal{L}_{\mathcal{B}}.$$

#### Initialize PINN model

Here, we define the residual form of the governing PDE, which is enforced inside the loss function of the PINN model.

 $$\mathcal{L}(\theta)=\underbrace{\langle{\left(\phi \frac{d\hat{S_{w}}}{dt}+v\frac{df_{w}(\hat{S_{w}})}{dx}\right)}\rangle^{2}}_{\mathcal{L}_{physics}}
    +\underbrace{{(\hat{S_{w}}(x,0)-S_{wr})}^{2}}_{\mathcal{L}_{I}}+\underbrace{{(\hat{S}_{w}(0,t)-(1-S_{nr}))}^{2}}_{\mathcal{L}_{B}},$$

```python
class PINN(tf.keras.Model):
    def __init__(self, neurons, layers, dim_in, dim_out):
        super(PINN, self).__init__() # Call methods from the mother class tf.keras.Model
        # Initial layer
        self.fci = tf.keras.layers.Dense(neurons, activation=None)
        # Output layer
        self.fco = tf.keras.layers.Dense(dim_out, activation=None)
        # Activation function
        self.tanh = tf.keras.layers.Activation('tanh')
        # Empty list for hidden layers
        self.mlps = []
        # Add hidden layers
        for _ in range(layers):
            self.mlps.append(tf.keras.layers.Dense(neurons, activation=None))
    
    def call(self, x, t):
        # Concatenate X and T
        x_i = tf.concat([x, t], axis=1)
        x_out = x_i # Create an output vector with the size of the input
        x_out = self.fci(x_out) # Evaluate x_i sequentaly using x_out
        
        for mlp_x in self.mlps:
            x_out = mlp_x(x_out)
            x_out = self.tanh(x_out)
        # Final x_out calculation
        x_out = self.fco(x_out)
        return x_out

# Physics loss function with entropy formulation
def physics_informed_loss_diff(model, bc_model, x, t, qt, phi, A):
    with tf.GradientTape(persistent=True) as tape:
        tape.watch(x)
        tape.watch(t)
        
        # Predict saturation
        sw_pred = model(x, t)
        
        # Fractional flow calculation with entropy fix
        fw = bc_model.fractional_flow(sw_pred)
        
        # Apply entropy fix: for S < sw_tangent, use linear fw = fw_swc + alpha * (S - swc)
        alpha = bc_model.alpha_tangent
        sw_tangent = bc_model.sw_tangent
        fw_swc = bc_model.fw_swc
        
        # Apply the entropy condition (shock capturing)
        fw = tf.where(sw_pred < sw_tangent, 
                      fw_swc + alpha * (sw_pred - swc),
                      fw)
    
    # Compute derivatives
    sw_t = tape.gradient(sw_pred, t)
    fw_x = tape.gradient(fw, x)
    
    del tape
    
    # Buckley-Leverett equation residual
    residual = sw_t + (qt / phi / A) * fw_x
    
    # Initial condition: S(x, 0) = swc
    t_zero = tf.zeros_like(t)
    sw_initial = model(x, t_zero)
    
    # Boundary condition: S(0, t) = 1-snwr
    x_zero = tf.zeros_like(x)
    sw_boundary = model(x_zero, t)
    
    # Residual forms of the loss funtions
    loss_physics = tf.reduce_mean(tf.abs(residual))
    loss_initial = tf.reduce_mean(tf.abs(sw_initial - swc))
    loss_boundary = tf.reduce_mean(tf.abs(sw_boundary - (1 - snwr)))
    
    total_loss = loss_physics + loss_boundary + loss_initial
    
    return total_loss, loss_physics, loss_boundary, loss_initial
```

```python
# Instantiate the model
print("Initializing PINN model...")
model = PINN(neurons=64, layers=5, dim_in=2, dim_out=1)
optimizer = tf.keras.optimizers.Adam(learning_rate=0.0005)

# Training
epochs_log = []
loss_totals = []
loss_physics_acc = []
loss_init_acc = []
loss_boundary_acc = []

num_epochs = 50000
print(f"\nStarting training for {num_epochs} epochs...\n")
starting_time = time.time()
for epoch in range(num_epochs):
    with tf.GradientTape() as tape:
        loss, loss_physics, loss_boundary, loss_initial = physics_informed_loss_diff(
            model, bc_model, X, T, qt, phi, A
        )
    
    gradients = tape.gradient(loss, model.trainable_variables)
    optimizer.apply_gradients(zip(gradients, model.trainable_variables))
    
    if epoch % 500 == 0 or epoch + 1 == num_epochs:
        print(f"Epoch {epoch}/{num_epochs}, Loss: {loss.numpy():.5f}, "
              f"Physics: {loss_physics.numpy():.5f}, "
              f"Initial: {loss_initial.numpy():.5f}, "
              f"Boundary: {loss_boundary.numpy():.5f}")
    
    epochs_log.append(epoch)
    loss_totals.append(loss.numpy())
    loss_physics_acc.append(loss_physics.numpy())
    loss_init_acc.append(loss_initial.numpy())
    loss_boundary_acc.append(loss_boundary.numpy())

print("\nTraining complete!")
print("Training time %1.5f"%((time.time()-starting_time)/60.0))
```
